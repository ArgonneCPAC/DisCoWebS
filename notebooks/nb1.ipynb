```python
from DisCoWebS.data_loader import clean_cosmos20_loader as cc20
from DisCoWebS.data_loader import sdss_loader as sdl
```

```python
cosmos = cc20.get_clean_cosmos20_data()
```

```python
sdss = sdl.load_sdss_wrapper()
```

```python
from DisCoWebS.data_loader import sdss_filter_select
sdss_filters_to_use = [
    "sdss_u","sdss_g","sdss_r","sdss_i"
]
tcurves, ssp_data, sdss_mag_colnames = sdss_filter_select.sdss_filters(sdss_filters_to_use)
```

```python
from DisCoWebS.optimizer import opt1
```

```python
params_bf_bounded, best_loss = opt1.run_optimization_1data(
    data=sdss,
    data_mag_colnames=sdss_mag_colnames,
    N_z_bins=4,
    N_host_min=20,
    N_host_max=30,
    n_z_phot_table=20,
    ssp_data=ssp_data,
    tcurves=tcurves,
    N_mag_bins=3,
    N_color_bins=4,
    ndsig_by_dbin=1.0,
    num_loops=3,
    num_steps_per_loop=3,
    modes=["fix_diffstarpop_merging",
           "fix_diffstarpop",
           "fit_all"],
    data_to_use="sdss",
    loss_type="ln_poisson_loss",
)
```

```python
import pprint

def namedtuple_to_dict(nt):
    """Recursively converts a nested named tuple to a nested dictionary."""
    d = {}
    for key in nt._fields:
        value = getattr(nt, key)
        if isinstance(value, tuple) and hasattr(value, '_asdict'):
            d[key] = namedtuple_to_dict(value)
        else:
            d[key] = value
    return d

all_pars = namedtuple_to_dict(params_bf_bounded)

pprint.pprint(all_pars)
```
