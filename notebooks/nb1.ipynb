```python
from DisCoWebS.data_loader import clean_cosmos20_loader as cc20
from DisCoWebS.data_loader import sdss_loader as sdl
```

```python
cosmos = cc20.get_clean_cosmos20_data()
```

```python
sdss = sdl.load_sdss_wrapper()
```

```python
from DisCoWebS.data_loader import cosmos_filter_select
cosmos_filters_to_use = [
    "g_HSC", "r_HSC", "i_HSC", "z_HSC", "y_HSC",
    "J_uv", "H_uv", "K_uv"
]
tcurves, ssp_data, cosmos_mag_colnames = cosmos_filter_select.cosmos_filters(cosmos_filters_to_use)
```

```python
from DisCoWebS.optimizer import opt1
```

```python
params_bf_bounded, best_loss = opt1.run_optimization(
    cosmos,
    cosmos_mag_colnames=cosmos_mag_colnames,
    N_z_bins=4,
    N_host_min=1000,
    N_host_max=1200,
    n_z_phot_table=30,
    ssp_data=ssp_data,
    tcurves=tcurves,
    N_mag_bins=6,
    N_color_bins=10,
    ndsig_by_dbin=0.5,
    num_loops=3,
    num_steps_per_loop=3,
    modes=["fix_diffstarpop_merging", "fix_diffstarpop", "fit_all"],
    data_to_use="cosmos"
)
```

```python
import pprint

def namedtuple_to_dict(nt):
    """Recursively converts a nested named tuple to a nested dictionary."""
    d = {}
    for key in nt._fields:
        value = getattr(nt, key)
        if isinstance(value, tuple) and hasattr(value, '_asdict'):
            d[key] = namedtuple_to_dict(value)
        else:
            d[key] = value
    return d

all_pars = namedtuple_to_dict(params_bf_bounded)

pprint.pprint(all_pars)
```
